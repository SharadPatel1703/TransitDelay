```python
import pandas as pd
from streamlit import columns
```

```python
df_weather = pd.read_csv('D:/(JSN)DataScience/Lectures/TransitDelayProj/filtered_files/weather_january.csv')
df_delay = pd.read_excel('D:/(JSN)DataScience/Lectures/TransitDelayProj/filtered_files/delay_january.xlsx')
weather = df_weather.copy()
delay = df_delay.copy()
print(df_weather.info())
print(df_delay.info())
```

```python
weather_new = weather.rename(columns={'Date/Time (LST)': 'datetime', 'Temp (°C)': 'temp', 'Precip. Amount (mm)': 'precipitation', 'Rel Hum (%)': 'humidity'})

delay_new = delay.rename(columns={'Min Delay': 'delay_minutes', 'Min Gap': 'gap_minutes'})

delay_new.columns = [col.lower() for col in delay_new.columns]

print(weather_new.info())
print(delay_new.info())
```

```python
weather_new[['date', 'time']] = weather_new['datetime'].str.split(' ', expand=True)

delay_new['date'] = delay_new['date'].dt.date
```

```python
weather_new.to_excel('new_weather.xlsx', index=False)
delay_new.to_excel('new_delay.xlsx', index=False)
```
