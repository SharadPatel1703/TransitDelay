```python
import pandas as pd
import numpy as np
```

```python
df_weather = pd.read_excel('D:/(JSN)DataScience/Lectures/TransitDelayProj/filtered_files/new_weather.xlsx')
df_delay = pd.read_excel('D:/(JSN)DataScience/Lectures/TransitDelayProj/filtered_files/new_delay.xlsx')

# print(df_weather)
# print(df_delay)
print(df_weather.info())
# print(df_delay.info())
```

```python
# Changing data type of Weather excel columns to needed ones

df_weather['datetime'] = pd.to_datetime(df_weather['datetime'])
df_weather['temp'] = df_weather['temp'].astype(float)
df_weather['precipitation'] = df_weather['precipitation'].astype(float)
df_weather['humidity'] = df_weather['humidity'].astype(float)

df_weather = df_weather.drop(['date', 'time'], axis=1)
print(df_weather.info())
# Done
```

```python
# Tasks:
# 1: Merge date and time column, drop old separated column
# 2: Change data types to the needed ones

# df_delay['date'] = df_delay['date'].astype(str)
# df_delay['time'] = df_delay['time'].astype(str)

# df_delay['datetime'] = df_delay['date'].fillna('') + ' ' + df_delay['time'].fillna('')
#
# df_delay['datetime'] = pd.to_datetime(df_delay['datetime'])
df_delay['route'] = pd.to_numeric(df_delay['route'], errors='coerce')
df_delay['day'] = df_delay['day'].str.lower().astype('string', errors='ignore')
df_delay['location'] = df_delay['location'].str.lower().astype(str, errors='ignore')
df_delay['incident'] = df_delay['incident'].str.lower().astype(str, errors='ignore')
df_delay['delay_minutes'] = pd.to_numeric(df_delay['delay_minutes'], errors='coerce')
df_delay['gap_minutes'] = pd.to_numeric(df_delay['gap_minutes'], errors='coerce')
df_delay['direction'] = df_delay['direction'].astype(str, errors='ignore').str.upper()
df_delay['vehicle'] = pd.to_numeric(df_delay['vehicle'], errors='coerce')

# df_delay = df_delay.drop(['date', 'time'], axis=1)

print(df_delay.info())
```

```python
# dropping null rows
df_delay = df_delay.dropna(subset=['route'])
df_delay['direction'] = df_delay['direction'].replace(['NAN', None, pd.NA,'nan'], 'unknown')
```

```python
df_delay.to_excel('delay_final.xlsx', index=False)
df_weather.to_excel('weather_delay.xlsx', index=False)
```
